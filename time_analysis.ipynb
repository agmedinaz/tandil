```python
def showEEG(signals,ti,tf,electrodes = [1,2,3,4,5,6,7,8,9]):

    fig = plt.figure(figsize=(10, 20))
    n=len(electrodes)
    j=0
    for i in electrodes:
        plt.subplot(n,1,j+1)
        plt.plot(signals[i][ti:tf] , color='purple' )
        plt.title('Ch: '+str(i+1))
        plt.ylim(-200, 200)
        j+=1
        #plt.xticks([])
    plt.show()
    return fig
```

```python
data.shape
```

```python
fig = showEEG(data, 0, 2560, range(21,24))
```
